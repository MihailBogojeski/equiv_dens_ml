```python
from pyscf import gto, dft
from pyscf.gto.mole import nao_nr
from pyscf.scf import hf
hf.MUTE_CHKFILE = True

mol = gto.M(atom='C  0  0  0.1184', basis='augccpvqzjkfit')
print('nprim', mol.bas_nprim(0))
print('nctr', mol.bas_nctr(0))
print('exp', mol.bas_exp(0))
print('kappa', mol.bas_kappa(0))
print('ctr_coeff', mol.bas_ctr_coeff(0))
print('libcint_ctr_coeff', mol._libcint_ctr_coeff(0))
print('env', mol._env)
print('bas', mol._bas)
print('atm', mol._atm)
print('basis', mol._basis)
print('nao_nr', nao_nr(mol))
mf = dft.RKS(mol)
mf.chkfile=False
mf.xc = 'pbe'
mf.kernel()
```

```python
mol = gto.M(atom='O  0  0  1; H  0,  0, 2; N 0,  0, 3; C 0, 0, 4; S 0, 0, 5; Cl 0 0 6; F 0 0 7', basis='augccpvqzjkfit')
ao_coeffs = {}
ao_basis = {}
repeat = {}
for i in range(mol._bas.shape[0]):
    a_ind = mol.bas_atom(i)
    a_num = mol._atm[a_ind, 0]
    symbol = chemical_symbols[a_num]
    if symbol not in ao_basis:
        ao_basis[symbol] = []
        ao_coeffs[symbol] = []
        repeat[symbol] = a_ind
    if repeat[symbol] != a_ind:
        continue
    L = mol.bas_angular(i)
    nprim = mol.bas_nprim(i)
    nctr = mol.bas_nctr(i)
    exp = mol.bas_exp(i)
    ctr = mol._libcint_ctr_coeff(i)
    for j in range(nctr):
        ao_basis[symbol].append((a_num, nprim, L))
        ao_coeffs[symbol].append((np.array(exp), np.array(ctr[:, j])))


print('ao basis', ao_basis)        
print('ao coeffs', ao_coeffs)
```

```python
print('ao_coeffs O', ao_coeffs['O'])
print('ao basis O', ao_basis['O'])
print('mol basis O', mol._basis['O'])
```

```python
np.save('datasets/augccpvqzjkfit_orbital_basis_libcint_df.npy', ao_basis, allow_pickle=True)
np.save('datasets/augccpvqzjkfit_radial_coeffs_libcint_df.npy', ao_coeffs, allow_pickle=True)
```
